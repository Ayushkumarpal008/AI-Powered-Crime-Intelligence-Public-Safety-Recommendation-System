```python
print(df["City"].unique())
print("\nTotal cities:", df["City"].nunique())
```

```python
ncr_cities = [
    "Delhi",
    "New Delhi",
    "Ghaziabad",
    "Noida",
    "Gurugram",
    "Gurgaon",
    "Faridabad"
]

ncr_df = df[df["City"].isin(ncr_cities)].copy()

print("Original dataset:", df.shape)
print("Delhi NCR dataset:", ncr_df.shape)

print("\nRecords by city:")
print(ncr_df["City"].value_counts())
```

```python
print("Total unique crime descriptions:")
print(ncr_df["Crime Description"].nunique())

print("\nTop crime types:")
print(ncr_df["Crime Description"].value_counts().head(10))
```
