```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/crime_dataset_india.csv")
print(df.shape)
```

```python
ncr_cities = [
    "Delhi",
    "Ghaziabad",
    "Faridabad"
]
df = df[df["City"].isin(ncr_cities)].copy()
print(df.shape)
```

```python
df["Weapon Used"] = df["Weapon Used"].fillna("Unknown")
```

```python
df["Date Reported"] = pd.to_datetime(
    df["Date Reported"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

df["Date of Occurrence"] = pd.to_datetime(
    df["Date of Occurrence"],
    format="%m-%d-%Y %H:%M",
    errors="coerce"
)

df["Time of Occurrence"] = pd.to_datetime(
    df["Time of Occurrence"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

df["Date Case Closed"] = pd.to_datetime(
    df["Date Case Closed"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)
```

```python
print(df[
    [
        "Date Reported",
        "Date of Occurrence",
        "Time of Occurrence",
        "Date Case Closed"
    ]
].isnull().sum())
```
