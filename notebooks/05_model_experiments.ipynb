# Prepare Data for Machine Learning

```python
# Make a copy of the dataframe
ml_df = df.copy()

# Drop datetime columns (already represented by engineered features)
ml_df = ml_df.drop(columns=[
    "Date Reported",
    "Date of Occurrence",
    "Time of Occurrence",
    "Date Case Closed"
])

# Separate features and target
X = ml_df.drop(columns=["Crime Description"])
y = ml_df["Crime Description"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

X.head()
```

```python
print("Categorical Columns:")
print(X.select_dtypes(include=["object", "category"]).columns.tolist())

print("\nNumerical Columns:")
print(X.select_dtypes(include=["int64", "int32", "float64"]).columns.tolist())

print("\nBoolean Columns:")
print(X.select_dtypes(include=["bool"]).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

# Remove ID column
X = X.drop(columns=["Report Number", "CaseResolutionDays"])

# Update feature lists
categorical_features = [
    "City",
    "Victim Gender",
    "Weapon Used",
    "Crime Domain",
    "Case Closed",
    "MonthName",
    "DayOfWeek",
    "VictimAgeGroup"
]

numerical_features = [
    "Crime Code",
    "Victim Age",
    "Police Deployed",
    "Year",
    "Month",
    "Day",
    "Hour"
]

boolean_features = [
    "IsWeekend",
    "IsNightCrime",
    "IsCaseClosed"
]

# Numerical pipeline
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

# Categorical pipeline
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing
preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numerical_features),
    ("cat", categorical_transformer, categorical_features),
    ("bool", "passthrough", boolean_features)
])

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set:", X_train.shape)
print("Testing Set :", X_test.shape)
```

```python
print(X.columns)
```

# Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Build the complete pipeline
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
rf_pipeline.fit(X_train, y_train)

# Predictions
y_pred = rf_pipeline.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix Shape:")
print(confusion_matrix(y_test, y_pred).shape)
```

# Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Decision Tree Pipeline
dt_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(
        random_state=42
    ))
])

# Train
dt_pipeline.fit(X_train, y_train)

# Predict
y_pred_dt = dt_pipeline.predict(X_test)

# Accuracy
dt_accuracy = accuracy_score(y_test, y_pred_dt)

print(f"Decision Tree Accuracy: {dt_accuracy:.4f}")

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_dt))

print("\nConfusion Matrix Shape:")
print(confusion_matrix(y_test, y_pred_dt).shape)
```

# Hyperparameter Tuning - Random Forest

```python
from sklearn.model_selection import GridSearchCV

# Random Forest Pipeline
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(random_state=42))
])

# Parameters to test
param_grid = {
    "classifier__n_estimators": [100, 200, 300],
    "classifier__max_depth": [10, 20, None],
    "classifier__min_samples_split": [2, 5],
    "classifier__min_samples_leaf": [1, 2]
}

# Grid Search
grid_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1,
    verbose=2
)

# Train
grid_search.fit(X_train, y_train)

# Best model
best_model = grid_search.best_estimator_

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross Validation Accuracy:")
print(grid_search.best_score_)
```

# Risk Level Prediction Model

```python
# Create ML dataset for Risk Level Prediction

risk_df = df.copy()

# Remove columns not available or causing target leakage
risk_df = risk_df.drop(columns=[
    "Crime Description",
    "Crime Domain",
    "CaseResolutionDays"
])

# Features and Target
X = risk_df.drop(columns=[
    "RiskLevel",
    "Report Number"
])

y = risk_df["RiskLevel"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nTarget Distribution:")
print(y.value_counts())
```

```python
print("Categorical Columns:")
print(X.select_dtypes(include=["object", "category", "string"]).columns.tolist())

print("\nNumerical Columns:")
print(X.select_dtypes(include=["int64", "int32", "float64"]).columns.tolist())

print("\nBoolean Columns:")
print(X.select_dtypes(include=["bool"]).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

categorical_features = [
    "City",
    "Victim Gender",
    "Weapon Used",
    "Case Closed",
    "MonthName",
    "DayOfWeek",
    "VictimAgeGroup"
]

numerical_features = [
    "Crime Code",
    "Victim Age",
    "Police Deployed",
    "Year",
    "Month",
    "Day",
    "Hour"
]

boolean_features = [
    "IsWeekend",
    "IsNightCrime",
    "IsCaseClosed"
]

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

risk_preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numerical_features),
    ("cat", categorical_transformer, categorical_features),
    ("bool", "passthrough", boolean_features)
])

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set:", X_train.shape)
print("Testing Set:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

risk_rf_pipeline = Pipeline([
    ("preprocessor", risk_preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ))
])

risk_rf_pipeline.fit(X_train, y_train)

y_pred_risk = risk_rf_pipeline.predict(X_test)

risk_accuracy = accuracy_score(y_test, y_pred_risk)

print(f"Risk-Level Random Forest Accuracy: {risk_accuracy:.4f}")

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_risk))
```
